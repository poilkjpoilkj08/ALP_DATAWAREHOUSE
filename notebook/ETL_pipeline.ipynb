```python
import pandas as pd
import os
from sqlalchemy import create_engine
from google.oauth2 import service_account

# 1. Autentikasi ke Google Cloud
# Path mapping ke docker
gcp_key_path = '/root/.gcp/bigquery-service-account.json'
credentials = service_account.Credentials.from_service_account_file(gcp_key_path)

project_id = 'afl3-dw-hotel' 

print("Library dan Autentikasi GCP berhasil di-load!")
```

```python
csv_files = {
    'hotels': 'data/raw/hotels.csv',
    'room_types': 'data/raw/room_types.csv',
    'booking_channels': 'data/raw/booking_channels.csv',
    'countries': 'data/raw/countries.csv'
}

for table_name, file_path in csv_files.items():
    print(f"Membaca {file_path}...")
    df = pd.read_csv(file_path)
    
    # Load ke bigquery dataset afl3_hotel_raw
    destination = f"afl3_hotel_raw.{table_name}"
    print(f"Loading data ke {destination}...")
    
    df.to_gbq(
        destination_table=destination,
        project_id=project_id,
        if_exists='replace', # kalo sudah ada buat ganti yang sudah ada sesuai namanya "replace"
        credentials=credentials
    )
    print(f"Tabel {table_name} sukses load\n")
```
